# 04 - Economic Impact Assessment
GDP drop and recovery, unemployment proxy, scatter analyses.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.covid_analyzer import COVIDAnalyzer
plt.style.use('seaborn-v0_8')
sns.set_context('talk')
```

```python
an = COVIDAnalyzer()
_ = an.load_data(True)
impact = an.analyze_economic_impact()
imp_df = pd.DataFrame(impact).T
imp_df.head()
```

## Initial Impact vs Recovery

```python
fig, ax = plt.subplots(figsize=(8,6))
ax.scatter(imp_df['initial_impact_pct'], imp_df['recovery_rate_pct'], alpha=0.6)
ax.set_xlabel('Initial Impact (%)')
ax.set_ylabel('Recovery Rate (%)')
ax.set_title('Economic Impact vs Recovery')
```
