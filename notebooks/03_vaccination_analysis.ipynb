# 03 - Vaccination Analysis and Effectiveness
Correlation of vaccination with cases/deaths; cross-country scatter and trends.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.covid_analyzer import COVIDAnalyzer
plt.style.use('seaborn-v0_8')
sns.set_context('talk')
```

```python
an = COVIDAnalyzer()
_ = an.load_data(True)
va = an.analyze_vaccination_effectiveness()
eff_df = pd.DataFrame(va).T
eff_df.head()
```

## Correlation Distribution

```python
fig, ax = plt.subplots(1,2, figsize=(14,5))
ax[0].hist(eff_df['correlation'].dropna(), bins=20); ax[0].set_title('Correlation with Cases')
ax[1].hist(eff_df['p_value'].dropna(), bins=20); ax[1].set_title('P-values')
plt.tight_layout()
```

## Top Countries by Significant Negative Correlation (protective effect)
